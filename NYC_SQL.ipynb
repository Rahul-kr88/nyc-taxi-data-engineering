```python
# NYC Taxi SQL Analytics

# Load enriched taxi data
taxi_df = spark.read.parquet(
    "/Volumes/workspace/nyc_taxi/raw_data/enriched_taxi_data"
)

# Create a temporary SQL view
taxi_df.createOrReplaceTempView("nyc_taxi_enriched")

# Verify the dataset
print("Total records:", taxi_df.count())

display(taxi_df.limit(5))
```

```python
hourly_metrics = spark.sql("""
SELECT
    pickup_hour,
    COUNT(*) AS total_trips,
    ROUND(SUM(total_amount), 2) AS total_revenue,
    ROUND(AVG(total_amount), 2) AS average_revenue
FROM nyc_taxi_enriched
WHERE pickup_date >= '2026-08-01'
  AND pickup_date < '2026-09-01'
GROUP BY pickup_hour
ORDER BY pickup_hour
""")

display(hourly_metrics)
```

```python
payment_metrics = spark.sql("""
SELECT
    payment_type,
    COUNT(*) AS total_trips,
    ROUND(SUM(total_amount), 2) AS total_revenue,
    ROUND(AVG(tip_amount), 2) AS average_tip
FROM nyc_taxi_enriched
WHERE pickup_date >= '2026-08-01'
  AND pickup_date < '2026-09-01'
GROUP BY payment_type
ORDER BY total_trips DESC
""")

display(payment_metrics)
```

```python
pickup_location_metrics = spark.sql("""
SELECT
    pickup_zone,
    pickup_borough,
    COUNT(*) AS total_trips,
    ROUND(SUM(total_amount), 2) AS total_revenue
FROM nyc_taxi_enriched
WHERE pickup_date >= '2026-08-01'
  AND pickup_date < '2026-09-01'
GROUP BY pickup_zone, pickup_borough
ORDER BY total_trips DESC
LIMIT 10
""")

display(pickup_location_metrics)
```

```python
dropoff_location_metrics = spark.sql("""
SELECT
    dropoff_zone,
    dropoff_borough,
    COUNT(*) AS total_trips,
    ROUND(SUM(total_amount), 2) AS total_revenue
FROM nyc_taxi_enriched
WHERE pickup_date >= '2026-08-01'
  AND pickup_date < '2026-09-01'
GROUP BY dropoff_zone, dropoff_borough
ORDER BY total_trips DESC
LIMIT 10
""")

display(dropoff_location_metrics)
```

```python
trip_duration_metrics = spark.sql("""
SELECT
    pickup_hour,
    COUNT(*) AS total_trips,
    ROUND(AVG(trip_duration_minutes), 2) AS avg_duration_minutes,
    ROUND(AVG(trip_distance), 2) AS avg_distance_miles
FROM nyc_taxi_enriched
WHERE pickup_date >= '2026-08-01'
  AND pickup_date < '2026-09-01'
GROUP BY pickup_hour
ORDER BY pickup_hour
""")

display(trip_duration_metrics)
```

```python
daily_revenue_metrics = spark.sql("""
SELECT
    pickup_date,
    COUNT(*) AS total_trips,
    ROUND(SUM(total_amount), 2) AS total_revenue,
    ROUND(AVG(total_amount), 2) AS average_fare
FROM nyc_taxi_enriched
WHERE pickup_date >= '2026-08-01'
  AND pickup_date < '2026-09-01'
GROUP BY pickup_date
ORDER BY pickup_date
""")

display(daily_revenue_metrics)
```

```python
borough_revenue_metrics = spark.sql("""
SELECT
    pickup_borough,
    COUNT(*) AS total_trips,
    ROUND(SUM(total_amount), 2) AS total_revenue,
    ROUND(AVG(total_amount), 2) AS average_revenue_per_trip
FROM nyc_taxi_enriched
WHERE pickup_date >= '2026-08-01'
  AND pickup_date < '2026-09-01'
GROUP BY pickup_borough
ORDER BY total_revenue DESC
""")

display(borough_revenue_metrics)
```

```python
borough_trip_percentage = spark.sql("""
SELECT
    pickup_borough,
    COUNT(*) AS total_trips,
    ROUND(
        COUNT(*) * 100.0 / SUM(COUNT(*)) OVER (),
        2
    ) AS trip_percentage
FROM nyc_taxi_enriched
WHERE pickup_date >= '2026-08-01'
  AND pickup_date < '2026-09-01'
GROUP BY pickup_borough
ORDER BY trip_percentage DESC
""")

display(borough_trip_percentage)
```

```python
august_summary = spark.sql("""
SELECT
    COUNT(*) AS total_trips,
    ROUND(SUM(total_amount), 2) AS total_revenue,
    ROUND(AVG(total_amount), 2) AS average_fare,
    ROUND(AVG(trip_distance), 2) AS average_distance_miles,
    ROUND(AVG(trip_duration_minutes), 2) AS average_duration_minutes,
    ROUND(AVG(tip_amount), 2) AS average_tip
FROM nyc_taxi_enriched
WHERE pickup_date >= '2026-08-01'
  AND pickup_date < '2026-09-01'
""")

display(august_summary)
```
